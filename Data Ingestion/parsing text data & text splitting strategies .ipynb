#### TextLoader - Read Single File


```python
from langchain.document_loaders import TextLoader
from langchain_community.document_loaders import TextLoader

#loading a single text file
loader=TextLoader("data/text_files/sample.txt",encoding="utf-8")

documents=loader.load()
print(type(documents))
print(documents)
```

This is of Document structure..even as we are loading it, it still contains metadata and page_Content in same format.

```python
print(f"Loaded {len(documents)} document")
print(f"Content preview : {documents[0].page_content[:100]}...")
print(f"Metadata: {documents[0].metadata}")
```

#### Directory Loader - Multiple Text Files

```python
from langchain_community.document_loaders import DirectoryLoader

#load all the text files
dir_loader=DirectoryLoader(
    "data/text_files",
    glob="**/*.txt", #  * - Folder...go to the folder and match any file having .txt
    loader_cls=TextLoader, #loader cls to use
    loader_kwargs={"encoding":"utf-8"},
    show_progress=True
)

documents=dir_loader.load()

print(f"Loaded {len(documents)} documents")

for i, doc in enumerate(documents):
    print(f"\nDocument {i+1}:")
    print(f" Source: {doc.metadata['source']}")
    print(f"Length: {len(doc.page_content)} characters")
```

glob - gives pattern and directns to check dirs and pattern to match files<br><br>
loader_cls usually stores a reference to a document loader class, not an instance.
It’s a convention used in LangChain when dynamically creating loaders.<br><br>
loader_kwargs → what extra options to give that loader.<br>
loader_kwargs lets you:
Pass optional settings dynamically.
Keep your loader configuration flexible.
Avoid hardcoding parameters.

Adv of directory loader:

Loads multiple files

supports glob patterns

progress tracking

recursive directory scanning
<br><br><br>

Disadv:

All files same type

Limited error handling

Can be hamemory intensive for large directories

#### Text Splitting Strategies

Docs are huge size, data is split into chunks. Every LLM has a context size, hence they are split into chunks.

```python
from langchain.text_splitter import(
    CharacterTextSplitter,
    RecursiveCharacterTextSplitter,
    TokenTextSplitter
)
print(documents)
```

METHOD -1 : CHARACTER TEXT SPLITTER

```python
text=documents[0].page_content
print(text)
```

```python
char_splitter=CharacterTextSplitter(
    separator="\n", #split on newlines
    chunk_size=200, #max chunk size in characters
    chunk_overlap=20, #overlap btw chunks, both chunks can have upto 20 same characters
    length_function=len #how to measure chunks size
)

char_chunks=char_splitter.split_text(text)
print(f"Created {len(char_chunks)} chunks")
print(f"First Chunk: {char_chunks[0][:100]}...")
```

```python
print(char_chunks[0])
print("-------------")
print(char_chunks[1])
print("--------------")
print(char_chunks[2])
```

#### Method 2 : Recursive Character Splitter (Recommended)

```python
recursvie_Splitter=RecursiveCharacterTextSplitter(
    separators=["\n\n","\n"," ",""],
    chunk_size=200,
    chunk_overlap=20,
    length_function=len
)

recursive_chunks=recursvie_Splitter.split_text(text)
print(f"Created {len(recursive_chunks)} chunks")
print(f"First Chunk: {recursive_chunks[0][:100]}...")
```

Overlapping of data in chunks is not necessarily a bad thing. When we do similarity search on the chunks, becuase of overlapping - both chunks will be pulled. We will have data from both chunks, hence getting better results for the search request.

#### Method 3 : Token-Based Splitter

```python
token_splitter=TokenTextSplitter(
    chunk_size=50, #size in tokens (not characters)
    chunk_overlap=10
)

token_chunks=token_splitter.split_text(text)
print(f"Created {len(token_chunks)} chunks")
print(f"First chunk: {token_chunks[0][:100]}...")
```

1. Character Splitter

Splits text by a fixed number of characters, regardless of meaning or sentence boundaries.

Simple and fast — good for plain, unstructured text.

May break sentences or lose context between chunks.

2. Recursive Character Splitter

Splits text hierarchically (paragraph → sentence → word → character) to keep natural structure.

Produces context-friendly chunks without cutting mid-sentence.

Ideal for structured data like PDFs, documentation, or RAG pipelines.

3. Token-Based Splitter

Splits text based on model tokens, not characters — matches how LLMs read text.

Ensures chunks stay within model context limits (e.g., 4K tokens).

Best for precise control and efficient LLM usage in retrieval or chat systems.
