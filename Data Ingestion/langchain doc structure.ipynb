```python
import os
from typing import List, Dict, Any 
import pandas as pd
```

```python
from langchain_core.documents import Document
from langchain.text_splitter import(
    RecursiveCharacterTextSplitter,
    CharacterTextSplitter,
    TokenTextSplitter
)
print("Setup completed")
```

Metadata is crucial for filtering search results, tracking document sources, providing context in responses, dubbing and auditing.

```python
#create a simple doc
doc=Document(
    page_content="This is the main text content to be searched and embedded.",
    metadata={
        "source":"example.txt",
        "page":1,
        "author":"Sahithi Akula",
        "date_created":"11/10/2025",
        "timestamp":"00:40"
    }
)
print("Doc Structure")

print(f"Content: {doc.page_content}")
print(f"Metadata: {doc.metadata}")

```

When we use langchain to load some pdf or any data, it will be loaded in Document format. The return type after it is loaded will be of class Document.
