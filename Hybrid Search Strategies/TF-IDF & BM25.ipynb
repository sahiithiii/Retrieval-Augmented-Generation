### TF-IDF

Title: TF-IDF (Term Frequency – Inverse Document Frequency)

TF-IDF is a method used in text retrieval to measure how important a word is in a document relative to a collection of documents.

1. Term Frequency (TF)
Term Frequency measures how often a word appears in a document.

TF = (Number of times the word appears in the document) / (Total number of words in the document)

2. Inverse Document Frequency (IDF)
Inverse Document Frequency indicates how unique or rare a word is across multiple documents.

IDF = log(Total number of documents / Number of documents containing the word)

Common words like "the", "is", "of" have low IDF, meaning they are less important.
Rare but meaningful words have high IDF, meaning they are more important.

3. TF-IDF Score
TF-IDF = TF × IDF

This score shows how important a word is within a specific document.

Example:
- The word "machine" appears 3 times in a document of 100 words:
  TF = 3 / 100 = 0.03

- The word "machine" appears in 5 out of 100 documents:
  IDF = log(100 / 5) = log(20)

- So, TF-IDF = 0.03 × log(20)

A higher TF-IDF score means the word is more important in that document.

### BM25 Algorithm

Title: BM25 Ranking Algorithm

BM25 is a ranking algorithm used in search engines to determine how relevant a document is to a user’s query. It is considered an improved and more accurate version of TF-IDF and is widely used in information retrieval systems such as Elasticsearch, Lucene, Vespa, Whoosh, and RAG pipelines.

Key Idea:
BM25 ranks documents based on three main factors:
1. How many times the query words appear in the document (Term Frequency).
2. How rare those words are across the entire document collection (Inverse Document Frequency).
3. Document length normalization, where shorter documents are preferred because they are typically more focused.

Why BM25 is Better Than TF-IDF:

| Feature                     | TF-IDF                       | BM25                                      |
|----------------------------|------------------------------|-------------------------------------------|
| Term frequency growth      | Linear                        | Saturates (repetitions do not infinitely increase importance) |
| Document length normalization | Weak                         | Strong (accounts for longer documents effectively) |
| Relevance performance      | Good                          | Better and widely used as the industry standard |

In BM25, repeating a word many times does not continually increase its weight. This prevents spam-like document structures where words are stuffed to manipulate ranking.

How BM25 Scoring Works (Conceptual):
For each word in the query, BM25 calculates its importance based on:
- How frequently the word appears in the document.
- How rare the word is in all documents.
- How long the document is compared to the average document length.

BM25 rewards:
- Words from the query that appear in the document.
- Words that are rare across documents.
- Short and focused documents.

BM25 penalizes:
- Very long documents that contain diluted information.
- Excessive repetition of the same word.

Example:
Query:
machine learning

Document 1: "machine learning basics"
Document 2: "football match highlights"

Document 1 receives a higher BM25 score because it contains the words "machine" and "learning," making it more relevant to the query. Document 2 does not contain these words, so its score is lower.

Where BM25 is Used:
- Search engines (Google-like search)
- RAG pipelines (retrieval for language models)
- Elasticsearch and OpenSearch systems
- Libraries such as Lucene, FAISS, and Haystack

In Short:
BM25 is a smarter version of TF-IDF that improves search relevance by focusing on meaningful word matches, reducing the impact of repeated terms, and adjusting for document length.
