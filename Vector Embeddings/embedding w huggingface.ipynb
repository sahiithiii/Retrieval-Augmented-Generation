### Embedding Models with HuggingFace

```python
#HuggingFace Models

from langchain_huggingface import HuggingFaceEmbeddings

#Initialize simple embedding models (No API Key needed since we have the library)
embeddings=HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
embeddings
```

U sed a sentence transformers. Sentence tranformers converts the entire sentence into vector embeddings, and not just words.

Converts the entire sentence into vector representation.

```python
text="Hello, I am learning about embeddings!" 


embedding=embeddings.embed_query(text)
print(text)
print(len(embedding))
print(embedding)
```

.embed_documents() is used for multiple queries

```python
sentences = [
    "The cat sat on the mat",
    "The cat sat on the mat",
    "The dog played in the yard",
    "I love programming in Python",
    "Python is my favorite programming language"
]

embedding_sen=embeddings.embed_documents(sentences)
print(len(embedding_sen))
print(embedding_sen)
```
